```python
import pandas as pd
import numpy as np

df = pd.read_csv("messy_customer_data.csv")

df.head()
```

```python
# Data Quality Report

print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nNumerical Summary:")
print(df.describe())
```

```python
# Check unique values in important columns

print("Gender Values:")
print(df["Gender"].unique())

print("\nCity Values:")
print(df["City"].unique())

print("\nAge Range:")
print(df["Age"].min(), "to", df["Age"].max())

print("\nPurchase Amount Range:")
print(df["Purchase_Amount"].min(), "to", df["Purchase_Amount"].max())

print("\nAnnual Income:")
print(df["Annual_Income"].unique())

print("\nJoin Date Values:")
print(df["Join_Date"].unique())
```

```python
# Make a copy before cleaning
cleaned_df = df.copy()

# Convert Age to numeric
cleaned_df["Age"] = pd.to_numeric(cleaned_df["Age"], errors="coerce")

# Convert Annual Income to numeric
cleaned_df["Annual_Income"] = (
    cleaned_df["Annual_Income"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)
cleaned_df["Annual_Income"] = pd.to_numeric(
    cleaned_df["Annual_Income"], errors="coerce"
)

# Convert Purchase Amount to numeric
cleaned_df["Purchase_Amount"] = pd.to_numeric(
    cleaned_df["Purchase_Amount"], errors="coerce"
)

# Convert Join Date to datetime
cleaned_df["Join_Date"] = pd.to_datetime(
    cleaned_df["Join_Date"], errors="coerce"
)

# Show missing values after type conversion
print("Missing values after type conversion:")
print(cleaned_df.isnull().sum())
```

```python
# Standardise Gender values

cleaned_df["Gender"] = (
    cleaned_df["Gender"]
    .astype(str)
    .str.strip()
    .str.lower()
)

cleaned_df["Gender"] = cleaned_df["Gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})

# Standardise City values

cleaned_df["City"] = (
    cleaned_df["City"]
    .astype(str)
    .str.strip()
    .str.title()
)

# Remove extra spaces from names

cleaned_df["Name"] = (
    cleaned_df["Name"]
    .astype(str)
    .str.strip()
)

print("Gender values:")
print(cleaned_df["Gender"].unique())

print("\nCity values:")
print(cleaned_df["City"].unique())
```

## Missing Data Handling

The following strategies were used to handle missing values:

- **Age:** Missing values are replaced with the median age because the median is less affected by abnormal age values.
- **Email:** Missing values are replaced with "Not Provided" because an email address cannot be reliably inferred.
- **Join_Date:** Invalid or missing dates are retained as missing (`NaT`) because an accurate date cannot be assumed.
- **Other numerical columns:** Missing values are handled using the median where required.

```python
# Handle missing values

# Age: replace missing values with median
cleaned_df["Age"] = cleaned_df["Age"].fillna(cleaned_df["Age"].median())

# Email: replace missing values with a text label
cleaned_df["Email"] = cleaned_df["Email"].fillna("Not Provided")

# Check remaining missing values
print("Missing values after handling:")
print(cleaned_df.isnull().sum())
```

```python
# Identify invalid ages

invalid_age = (cleaned_df["Age"] < 18) | (cleaned_df["Age"] > 100)

print("Invalid ages:")
print(cleaned_df.loc[invalid_age, ["Customer_ID", "Name", "Age"]])
```

```python
# Replace invalid ages with NaN
cleaned_df.loc[invalid_age, "Age"] = np.nan

# Fill invalid ages using the median age
cleaned_df["Age"] = cleaned_df["Age"].fillna(cleaned_df["Age"].median())

print("Age range after cleaning:")
print(cleaned_df["Age"].min(), "to", cleaned_df["Age"].max())
```

```python
# Detect outliers using IQR

Q1 = cleaned_df["Purchase_Amount"].quantile(0.25)
Q3 = cleaned_df["Purchase_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = cleaned_df[
    (cleaned_df["Purchase_Amount"] < lower_limit) |
    (cleaned_df["Purchase_Amount"] > upper_limit)
]

print("\nOutliers:")
print(outliers[["Customer_ID", "Name", "Purchase_Amount"]])
```

```python
# Count duplicates before removal

duplicates_before = cleaned_df.duplicated().sum()

print("Duplicate rows before removal:", duplicates_before)

# Remove duplicate rows
cleaned_df = cleaned_df.drop_duplicates()

# Count duplicates after removal
duplicates_after = cleaned_df.duplicated().sum()

print("Duplicate rows after removal:", duplicates_after)
```

```python
# Final Data Quality Check

print("Final Dataset Shape:")
print(cleaned_df.shape)

print("\nMissing Values:")
print(cleaned_df.isnull().sum())

print("\nDuplicate Rows:")
print(cleaned_df.duplicated().sum())

print("\nFinal Data Types:")
print(cleaned_df.dtypes)

print("\nFirst 5 Cleaned Records:")
display(cleaned_df.head())
```
