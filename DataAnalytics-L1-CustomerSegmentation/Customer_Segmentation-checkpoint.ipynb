```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("store_customers.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Column names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Annual Income (k$)'] = pd.to_numeric(
    df['Annual Income (k$)'], errors='coerce'
)
df['Spending Score (1-100)'] = pd.to_numeric(
    df['Spending Score (1-100)'], errors='coerce'
)

df.dropna(inplace=True)

print("Shape after cleaning:", df.shape)
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='Gender')

plt.title('Customer Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['Age'], bins=20, kde=True)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['Annual Income (k$)'], bins=20, kde=True)

plt.title('Annual Income Distribution')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['Spending Score (1-100)'], bins=20, kde=True)

plt.title('Spending Score Distribution')
plt.xlabel('Spending Score')
plt.ylabel('Number of Customers')

plt.show()
```

```python
features = ['Annual Income (k$)', 'Spending Score (1-100)']

X = df[features]

X.head()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Data standardized successfully!")
```

```python
inertia = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(range(1, 11), inertia, marker='o')

plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

df['Cluster'] = kmeans.fit_predict(X_scaled)

print("K-Means clustering completed!")
```

```python
df.head(10)
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Cluster',
    s=100
)

plt.title('Customer Segmentation using K-Means')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')

plt.show()
```

```python
centers = scaler.inverse_transform(kmeans.cluster_centers_)

centers_df = pd.DataFrame(
    centers,
    columns=features
)

centers_df
```

```python
cluster_summary = df.groupby('Cluster')[features].mean().round(2)

cluster_summary
```

```python
cluster_size = df['Cluster'].value_counts().sort_index()

print("Customers in each cluster:")
print(cluster_size)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Cluster')

plt.title('Number of Customers in Each Cluster')
plt.xlabel('Cluster')
plt.ylabel('Number of Customers')

plt.show()
```

```python
gender_cluster = pd.crosstab(
    df['Cluster'],
    df['Gender']
)

gender_cluster
```

```python
df.to_csv('customer_segments.csv', index=False)

print("Customer segmentation results saved successfully!")
```

## Conclusion

K-Means clustering was used to segment customers based on their Annual Income and Spending Score.

The Elbow Method was used to determine a suitable number of clusters. Five clusters were selected for the final model.

The segmentation helps identify different groups of customers based on their purchasing behaviour. These customer groups can help businesses design targeted marketing strategies, personalized offers, and better customer engagement.

Overall, customer segmentation provides useful insights into customer behaviour and can support data-driven business decisions.
